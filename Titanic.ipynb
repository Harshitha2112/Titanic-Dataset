```python
"""
sum(data.isnull())
This uses Python’s built-in sum(), not pandas.
sum(data.isnull()) → manually stacking columns and adding them
data.isnull().sum() → pandas doing a proper column-wise calculation
   A      B
0  True   False
1  False  True
2  True   True

sum(data.isnull())
treats --> column A + column B
adds them like -->[True, False, True] +[False, True, True]
which becomes -->[1+0, 0+1, 1+1] = [1,1,2]
Now instead of column counts, you get row-wise mixed results

data.isnull().sum()	counts per column ✅
sum(data.isnull())	adds columns together ❌

Use Python sum() for:

lists
tuples
simple iterables

👉 Use pandas .sum() for:

Series
DataFrames
"""


data.isnull().sum()
```

<span style="font-size:20px">**Handle missing values**</span>

```python
# which o age and embarked column
data.columns
```

```python
#checking most frequently appearing datavalue

data['Embarked'].mode()
```

```python
#missing values are filled with most frequently occuring value that is S
#this works for categorical datavalues
data['Embarked'].fillna('S',inplace=True)
data.isnull().sum() #now there is no null value in embarked
```

```python
#It has numerical values
data['Age']
```

```python
#so we are going fill the missing values with the mean of age column
data['Age'].fillna(data['Age'].mean(),inplace=True)
data.isnull().sum()
```

<span style="font-size:20px">**Categorical Data Encoding**</span>

```python
#Encoding = convert categorical data → numerical data
#Because models understand numbers, not text
#A machine learning model cannot understand these strings
"""Because models do math like:

addition
multiplication
distance calculation

👉 Strings like 'S' + 'C' make no sense to a model"""

data.columns
```

```python
#checkong what are all categories are present
data['Sex'].unique()
```

```python
#mapping numerical values to categorical values
data["Gender"]=data['Sex'].map({'male':1,'female':0})
data.drop('Gender_new',axis=1,inplace=True)

data.head(1)
```

```python
#here gender column added at the end but i need it to be next to sex 
#insert(loc,col_name,which column to assign(values))
x=data['Sex'].map({'male':1,'female':0})
data.insert(5,'Gender_new',x)
```
